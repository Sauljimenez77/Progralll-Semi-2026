```python
# Función anónima para calcular el salario bruto

salario = lambda horas, valor: horas * valor * 30

horas = float(input("Horas trabajadas por día: "))
valor = float(input("Valor por hora: "))

bruto = salario(horas, valor)

# Descuentos
isss = bruto * 0.03
afp = bruto * 0.0725

# Base para calcular ISR
base = bruto - isss - afp

# Cálculo del ISR según la tabla mensual
if base <= 550.00:
    isr = 0
elif base <= 895.24:
    isr = ((base - 550.00) * 0.10) + 17.67
elif base <= 2038.10:
    isr = ((base - 895.24) * 0.20) + 60.00
else:
    isr = ((base - 2038.10) * 0.30) + 288.57

neto = base - isr

print("\n------ DESPRENDIBLE DE PAGO ------")
print("Salario Bruto: $", round(bruto, 2))
print("ISSS (3%): $", round(isss, 2))
print("AFP (7.25%): $", round(afp, 2))
print("ISR: $", round(isr, 2))
print("Salario Neto: $", round(neto, 2))
```

```python
#FILTER
#dada una serie numero dame solo lso que son pares
serie = [2,5,6,8,9,10,12]

pares = filter(lambda num: num%2==0, serie)
impares = filter(lambda num: num%2!=0, serie)
print("Original: ", serie)
print("Pares: ", list(pares))
print("Impares: ", list(impares))
```

```python
#Hacer en python el siguiente ejercicio, haciendo uso de filter y lambda.
#La notas introducidas por el usuario clasificarlas en aprobadas y reprobadas
notas = input("Notas: ") #5,6,1,8,9
notas = notas.split(",")
aprobados = filter(lambda nota: float(nota)>=6, notas)
reprobados = filter(lambda nota: float(nota)<6, notas)
print("Notas: ", notas)
print("Aprobados: ", list(aprobados))
print("Reprobados: ", list(reprobados))
```

```python

#reduce
#Cuanto es el acumulado de los numeros pares de una serie
from functools import reduce

numeros = [1,2,8,4,9,6]
pares = filter(lambda num: num%2==0, numeros)

print("Numeros:" , numeros)
#print("Pares: ", list(pares))

sumaPares = reduce(lambda acum, num: num+acum, pares, 0)
print("Suma de pares: ", sumaPares)
```
